```python
import PyPDF2
# import json

def pdf_to_txt_with_pypdf2(pdf_file, out_file):
    f = open(out_file, "a")    
 
    with open(pdf_file, 'rb') as file:
        reader = PyPDF2.PdfReader(file)
        data = {}
        
        # Loop through each page
        for i in range(len(reader.pages)):
            page = reader.pages[i]
            text = page.extract_text()
            print(text)
            f.write(text)
    file.close()
    f.close()



pdf_file = "../../data/raw/EngineeringReference.pdf"
out_file = "../../data/processed/EngineeringReference.txt"
pdf_to_txt_with_pypdf2(pdf_file, out_file)
```
